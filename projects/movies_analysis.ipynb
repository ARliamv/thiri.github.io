```python
# Import libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_style("whitegrid")
plt.rcParams["figure.figsize"] = (10,6)
```

```python
# Load dataset
df = pd.read_csv("movies_dataset.csv")

# Display first five rows
df.head()
```

```python
# Shape of dataset
print("Dataset Shape:", df.shape)
```

```python
# Column names
df.columns
```

```python
# Information  
df.info() 
```

```python
# Summary statistics
df.describe()
```

```python
# Data types
df.dtypes
```

```python
# Missing values
df.isnull().sum()
```

```python
#Convert Release Date
df["ReleaseDate"] = pd.to_datetime(
    df["ReleaseDate"],
    format="%d-%m-%Y",
    errors="coerce"
)
```

```python
df.isnull().sum() #missing values
```

```python
# Remove rows with missing important values
df = df.dropna(subset=[
    "BudgetUSD",
    "Global_BoxOfficeUSD",
    "IMDbRating"
])
```

```python
#Check Duplicates
df.duplicated().sum()
```

```python
df = df.drop_duplicates()
```

```python
# Check dataset size after cleaning
print("Dataset Shape After Cleaning:", df.shape)

# Show first 5 rows after cleaning
df.head()
```

```python
#Create profit
df["Profit"] = df["Global_BoxOfficeUSD"] - df["BudgetUSD"]
```

```python
df = df[df["BudgetUSD"] > 0]

df["ROI"] = (df["Profit"] / df["BudgetUSD"]) * 100
```

```python
#Create ROI
df["ROI"] = (df["Profit"] / df["BudgetUSD"]) * 100
```

```python
#correlation analysis
corr = df[
    [
        "BudgetUSD",
        "Global_BoxOfficeUSD",
        "Profit",
        "ROI",
        "IMDbRating",
        "RottenTomatoesScore"
    ]
].corr()

#correlation heatmap
plt.figure(figsize=(8,6))

sns.heatmap(
    corr,
    annot=True,
    cmap="Blues"
)

plt.title("Correlation Between Movie Variables")

plt.show()
```

```python
#Movie Production Over Time
movies_per_year = df.groupby("ReleaseYear").size()

movies_per_year.plot(
    kind="line",
    marker="o"
)

plt.title("Movies Released Each Year")
plt.xlabel("Year")
plt.ylabel("Number of Movies")
plt.show()
```

```python
#Average Box Office Over Time
boxoffice = df.groupby("ReleaseYear")["Global_BoxOfficeUSD"].mean()

boxoffice.plot()

plt.title("Average Global Box Office by Year")
plt.ylabel("Revenue")
plt.show()
```

```python
#IMDb Rating Over Time
rating = df.groupby("ReleaseYear")["IMDbRating"].mean()

rating.plot(color="green")

plt.title("Average IMDb Rating by Year")
plt.ylabel("IMDb Rating")
plt.show()
```

```python
#Budget vs Revenue
sns.scatterplot(
    data=df,
    x="BudgetUSD",
    y="Global_BoxOfficeUSD"
)

plt.title("Budget vs Revenue")
plt.show()
print("The plot suggests a positive relationship between budget and global box office revenue, meaning movies with larger budgets often achieve higher revenues, although exceptions exist.")
```

```python
#top 10 movies
top_movies = df.nlargest(
    10,
    "Global_BoxOfficeUSD"
)

# Plot top 10 
plt.figure(figsize=(10,6))

sns.barplot(
    data=top_movies,
    x="Global_BoxOfficeUSD",
    y="Title"
)

plt.title("Top 10 Highest Grossing Movies")
plt.xlabel("Worldwide Revenue (USD)")
plt.ylabel("Movie Title")

plt.show()
```

```python
#Budget vs ROI
sns.scatterplot(
    data=df,
    x="BudgetUSD",
    y="ROI"
)

plt.title("Budget vs ROI")
plt.show()
print("The relationship between budget and ROI appears weak, suggesting that a higher budget does not always guarantee a better return on investment.")
```

```python
#Correlation
budget_revenue = df["BudgetUSD"].corr(df["Global_BoxOfficeUSD"])
budget_roi = df["BudgetUSD"].corr(df["ROI"])

print("Budget & Revenue:", budget_revenue)
print("Budget & ROI:", budget_roi)
```

```python
#Split genres
genre_df = df.assign(
    Genre=df["Genre"].str.split("|")
).explode("Genre")
```

```python
genre_profit = (
    genre_df.groupby("Genre")["Profit"]
    .mean()
    .sort_values(ascending=False)
)

genre_profit.head(10).plot(kind="bar")

plt.title("Average Profit by Genre")
plt.ylabel("Average Profit")
plt.show()
```

```python
#Most Popular Genres
popular = genre_df["Genre"].value_counts()

popular.plot(kind="bar")

plt.title("Most Popular Movie Genres")
plt.show()
```

```python
#Genre Trends Over Time
genre_year = genre_df.groupby(
    ["ReleaseYear","Genre"]
).size().unstack(fill_value=0)

genre_year.plot(figsize=(12,6))

plt.title("Genre Trends Over Time")
plt.show()
```

```python
# average rating for each genre
genre_rating = (
    genre_df.groupby("Genre")["IMDbRating"]
    .mean()
    .sort_values(ascending=False)
)


plt.figure(figsize=(10,6))

sns.barplot(
    x=genre_rating.values,
    y=genre_rating.index
)

plt.title("Average IMDb Rating by Genre")
plt.xlabel("Average IMDb Rating")
plt.ylabel("Genre")

plt.show()
```

```python
#Filter directors with more than 3 movies
director = df.groupby("Director").agg(
    Movies=("Title","count"),
    Avg_Rating=("IMDbRating","mean"),
    Avg_Profit=("Profit","mean")
)

director = director[director["Movies"] > 3]
```

```python
#Sort by Rating
director.sort_values(
    by="Avg_Rating",
    ascending=False
).head(10)
```

```python
#Sort by Profit
director.sort_values(
    by="Avg_Profit",
    ascending=False
).head(10)
```

```python
#Visualization
top_director = director.sort_values(
    by="Avg_Rating",
    ascending=False
).head(10)

sns.barplot(
    data=top_director,
    x=top_director.index,
    y="Avg_Rating"
)

plt.xticks(rotation=45)
plt.title("Top Directors by IMDb Rating")
plt.show()
```
